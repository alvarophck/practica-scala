## Ejercicio 2 — val, var y reasignación

```scala
val jugador = "Marta"
var puntuacion = 10

puntuacion = puntuacion + 5
puntuacion = puntuacion + 3
println(s"Puntuación final: $puntuacion")
```

```scala
jugador = "Sindhu"
```

`puntuacion` puede modificarse porque se declaró con `var`, que permite reasignación.
`jugador` no puede reasignarse porque se declaró con `val`, que crea una referencia inmutable.

## Ejercicio 3 — Tipos numéricos y precisión

## Ejercicio 4 — Función para determinar si una mano se pasa de 21

```scala
def bust(puntuacion: Int): Boolean = {
  puntuacion > 21
}

println(bust(18))
println(bust(21))
println(bust(22))
println(bust(30))
```

## Ejercicio 5 — Comparación de dos manos

```scala
def maxHand(handA: Int, handB: Int): Int = {
  if (handA > handB) handA
  else handB
}

println(maxHand(17, 19))
println(maxHand(20, 18))
println(maxHand(21, 21))
```

Cuando ambas manos tienen el mismo valor, la función devuelve `handB` porque la condición `handA > handB` es falsa y se ejecuta el `else`. El resultado es correcto (devuelve el valor empatado), aunque técnicamente "elige" siempre `handB` en caso de empate. No he modificado la función porque el comportamiento sigue siendo válido: cuando hay empate, da igual cuál de las dos se devuelva, ya que ambas tienen el mismo valor.

## Ejercicio 6 — Decidir el ganador de una partida

```scala
def ganador(handA: Int, handB: Int): Int = {
  if (bust(handA) && bust(handB)) 0
  else if (bust(handA)) handB
  else if (bust(handB)) handA
  else maxHand(handA, handB)
}

println(ganador(26, 20))
println(ganador(18, 22))
println(ganador(24, 25))
println(ganador(17, 19))
println(ganador(21, 20))
```

- ganador(26, 20) → 0: ambas manos superan 21 (primera condición)
- ganador(18, 22) → 18: solo handB supera 21 (tercera condición)
- ganador(24, 25) → 0: ambas manos superan 21 (primera condición)
- ganador(17, 19) → 19: ninguna supera 21, se usa maxHand (última condición)
- ganador(21, 20) → 21: ninguna supera 21, se usa maxHand (última condición)

## Ejercicio 7 — Arrays y mutabilidad

```scala
val jugadores = Array("Alex", "Chen", "Marta")
println(jugadores.mkString(", "))

jugadores(0) = "Sindhu"
println(jugadores.mkString(", "))
```

```scala
jugadores(1) = 500
```

Los elementos del array pueden cambiar aunque `jugadores` se haya declarado con `val`, porque `val` solo impide reasignar la variable a un array distinto, pero no impide modificar los elementos internos del array (el array en sí es una estructura mutable).

Scala no permite introducir el valor `500` en un `Array[String]` porque el array tiene un tipo fijo definido en el momento de su creación (inferido como `String` a partir de sus elementos iniciales), y `500` es un `Int`, no un `String`.

La diferencia es que reasignar la variable cambiaría a qué array completo apunta `jugadores` (no permitido con `val`), mientras que modificar un elemento cambia solo una posición dentro del mismo array (sí permitido, porque el array es mutable internamente).

## Ejercicio 8 — Creación e inicialización de Arrays

```scala
val puntuaciones = new Array[Int](4)
println(puntuaciones.mkString(", "))

puntuaciones(0) = 17
puntuaciones(1) = 24
puntuaciones(2) = 21
puntuaciones(3) = 19

println(puntuaciones.mkString(", "))
println(puntuaciones.length)
```

Un `Array[Int]` recién creado contiene el valor por defecto `0` en todas sus posiciones, ya que Scala inicializa los arrays numéricos con el valor neutro de su tipo en lugar de dejarlos vacíos o nulos.

## Ejercicio 9 — Recorrer un Array con while

```scala
val manos = Array(17, 24, 21, 19, 26)

var i = 0
while (i < manos.length) {
  println(s"Mano: ${manos(i)}, ¿Se pasa? ${bust(manos(i))}")
  i = i + 1
}
```

El contador `i` debe ser mutable (`var`) porque necesita incrementarse en cada vuelta del bucle para avanzar por las posiciones del array; con `val` no se podría modificar tras la primera asignación.

Si no incrementara `i`, el bucle se ejecutaría infinitamente, ya que la condición `i < manos.length` nunca dejaría de cumplirse.

Este ejercicio corresponde al estilo imperativo: se usa una variable mutable como contador, se controla explícitamente el flujo de ejecución con `while`, y se depende del estado externo (`i`) para decidir cuándo parar.

## Ejercicio 10 — Listas e inmutabilidad

```scala
val jugadoresLista = List("Alex", "Chen", "Marta")
println(jugadoresLista)

val jugadoresConSindhu = "Sindhu" :: jugadoresLista
println(jugadoresConSindhu)
println(jugadoresLista)

println(jugadoresLista.length)
println(jugadoresLista.reverse)
```

La operación `::` produce una lista nueva en lugar de modificar la lista original porque `List` en Scala es una estructura inmutable: cualquier operación que "añada" un elemento en realidad construye una lista completamente nueva, reutilizando internamente los elementos de la lista original, pero sin alterarla.

## Ejercicio 11 — Construcción y concatenación de listas

```scala
val listaUno = "Ana" :: "Luis" :: "Marta" :: Nil
val listaDos = "Pedro" :: "Sofia" :: Nil

val listaCombinada = listaUno ::: listaDos

println(listaUno)
println(listaDos)
println(listaCombinada)
```

Como se observa en la salida, `listaUno` y `listaDos` permanecen exactamente igual después de la concatenación con `:::`, ya que esta operación crea una tercera lista nueva (`listaCombinada`) sin modificar ninguna de las dos originales, respetando la inmutabilidad de `List`.

## Ejercicio 12 — Operadores relacionales y lógicos

```scala
val handA = 18
val handB = 21
val handC = 25

println(handA > handB)
println(handB == 21)
println(handC != 21)
println(handA <= 21 && handB <= 21)
println(handA > 21 || handC > 21)
println(!(handB > 21))
```

| Expresión | Resultado | Explicación |
|---|---|---|
| `handA > handB` | false | 18 no es mayor que 21 |
| `handB == 21` | true | 21 es igual a 21 |
| `handC != 21` | true | 25 es distinto de 21 |
| `handA <= 21 && handB <= 21` | true | ambas condiciones son ciertas (18 y 21 son ≤ 21) |
| `handA > 21 || handC > 21` | true | la segunda condición es cierta (25 > 21) |
| `!(handB > 21)` | true | `handB > 21` es falso, y la negación de falso es verdadero |

## Ejercicio 13 — foreach y funciones como valores

```scala
val manosEj13 = Array(17, 24, 21, 26, 18)

var j = 0
while (j < manosEj13.length) {
  println(s"Mano: ${manosEj13(j)}, ¿Se pasa? ${bust(manosEj13(j))}")
  j = j + 1
}
```

```scala
manosEj13.foreach(mano => println(s"Mano: $mano, ¿Se pasa? ${bust(mano)}"))
```

La versión con `while` necesita un contador (`j`) para saber en qué posición del array está y cuándo detenerse, además de una variable mutable (`var`) para ir avanzando ese contador.

La versión con `foreach` no necesita ni contador ni variable mutable: recibe una función que se aplica automáticamente a cada elemento de la colección.

La versión con `foreach` se aproxima más al estilo funcional presentado en el material del curso, ya que se centra en "qué hacer con cada elemento" en lugar de "cómo recorrer manualmente la colección".

## Ejercicio 14 — Efectos secundarios y estilo de programación

```scala
var total = 0

def sumarAlTotal(valor: Int) = {
  total = total + valor
}

sumarAlTotal(10)
println(s"Total tras primera llamada: $total")

sumarAlTotal(5)
println(s"Total tras segunda llamada: $total")

sumarAlTotal(20)
println(s"Total tras tercera llamada: $total")
```

```scala
def sumar(a: Int, b: Int): Int = {
  a + b
}

println(sumar(10, 5))
println(sumar(20, 7))
```

`sumarAlTotal` modifica la variable `total`, que está definida fuera de su ámbito local, porque Scala permite que las funciones accedan y modifiquen variables `var` del ámbito que las contiene (closures). El efecto secundario es precisamente esa modificación de `total`, un estado externo a la función, en lugar de limitarse a devolver un resultado.

| Característica | `sumarAlTotal` | `sumar` |
|---|---|---|
| Modifica datos externos | Sí | No |
| Devuelve un resultado calculado | No (devuelve Unit) | Sí |
| Utiliza efecto secundario | Sí | No |
| Estilo predominante | Imperativo | Funcional |

## Ejercicio 15 — Programa integrado: torneo de Twenty-One

```scala
val jugadoresFinal = List("Alex", "Chen", "Marta", "Sindhu")
val manosRonda1 = Array(18, 24, 21, 20)

def evaluarRonda(nombres: List[String], puntos: Array[Int]): Unit = {
  var indice = 0
  while (indice < puntos.length) {
    val nombreJugador = nombres(indice)
    val puntuacionJugador = puntos(indice)
    val sePasa = bust(puntuacionJugador)
    println(s"$nombreJugador: $puntuacionJugador -> ¿Se pasa? $sePasa")
    indice = indice + 1
  }
}

evaluarRonda(jugadoresFinal, manosRonda1)

def mejorPuntuacionValida(puntos: Array[Int]): Int = {
  var mejor = 0
  puntos.foreach { puntuacion =>
    if (!bust(puntuacion) && puntuacion > mejor) {
      mejor = puntuacion
    }
  }
  mejor
}

println(s"Mejor puntuación válida ronda 1: ${mejorPuntuacionValida(manosRonda1)}")
```

```scala
val manosRonda2 = Array(22, 19, 20, 21)

evaluarRonda(jugadoresFinal, manosRonda2)
println(s"Mejor puntuación válida ronda 2: ${mejorPuntuacionValida(manosRonda2)}")
```

### Informe final del ejercicio

- **Mutable**: la variable `indice` dentro de `evaluarRonda` y la variable `mejor` dentro de `mejorPuntuacionValida`, ambas usadas como acumuladores/contadores internos.
- **Inmutable**: `jugadoresFinal` (List), `manosRonda1` y `manosRonda2` (Array, aunque el array en sí permite mutación interna, aquí no se modifican sus elementos), y todos los parámetros de las funciones.
- **Funciones**: `bust`, `evaluarRonda`, `mejorPuntuacionValida`.
- **Estructuras de control**: `while` dentro de `evaluarRonda`, `if` dentro de `mejorPuntuacionValida`.
- **Estilo funcional**: el uso de `foreach` en `mejorPuntuacionValida`, y el hecho de que ambas funciones reciben datos y devuelven resultados sin modificar las listas/arrays originales.
- **Estilo imperativo**: el uso de `while` con contador en `evaluarRonda`, y el uso de una variable `var` acumuladora en `mejorPuntuacionValida`.
